這個支線練習資料載入與初步分析，但可用資料只有兩類：

```python
import pandas as pd, numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_squared_error, roc_auc_score
```

### 步驟 1　載入資料，看一眼

```python
# 本支線只使用課程固定 seed 合成資料；不要上傳本機檔案或掛載雲端硬碟。
df = pd.read_csv("https://raw.githubusercontent.com/johnnychao/stats-quest-2026/v1.2.0/data/public/auto.csv")
print(df.shape); print(df.dtypes); df.describe(include="all").T.head(15)
```

### 步驟 2　決定 Y 與 X，做 one-hot，切分

```python
目標 = "mpg"                                   # ← 改成你要預測的欄位
是分類 = df[目標].nunique() <= 10             # 類別很少就當分類；不對的話自己改成 True / False
丟掉 = ["name"]                                # ← 文字型、編號型的欄位丟掉
X = pd.get_dummies(df.drop(columns=[目標] + 丟掉), drop_first=True).astype(float)
y = df[目標]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y if 是分類 else None)
print("是分類？", 是分類, "| 特徵數", X.shape[1], "| 訓練", len(X_train), "測試", len(X_test))
```

### 步驟 3　基線模型與評估（比瞎猜好多少？）

```python
if 是分類:
    基線 = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(X_train, y_train)
    進階 = RandomForestClassifier(300, random_state=42).fit(X_train, y_train)
    print("基線 AUC", round(roc_auc_score(y_test, 基線.predict_proba(X_test)[:, 1]), 3), "| 隨機森林 AUC", round(roc_auc_score(y_test, 進階.predict_proba(X_test)[:, 1]), 3), "| 瞎猜 0.5")
else:
    基線 = LinearRegression().fit(X_train, y_train)
    進階 = RandomForestRegressor(300, random_state=42).fit(X_train, y_train)
    瞎猜 = np.sqrt(mean_squared_error(y_test, np.full(len(y_test), y_train.mean())))
    print("基線 RMSE", round(np.sqrt(mean_squared_error(y_test, 基線.predict(X_test))), 3), "| 隨機森林 RMSE", round(np.sqrt(mean_squared_error(y_test, 進階.predict(X_test))), 3), "| 猜平均的 RMSE", round(瞎猜, 3))
```

### 步驟 4　哪個線索最重要？

```python
重要性 = pd.Series(進階.feature_importances_, index=X.columns).sort_values(ascending=False)
print(重要性.round(3).head(8))
```
